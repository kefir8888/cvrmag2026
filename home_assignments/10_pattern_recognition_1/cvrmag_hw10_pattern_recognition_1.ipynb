### Задача 2

Реализуйте преобразование Хафа для прямых. Делать это нужно самостоятельно, без дипсика, чат гпт и так далее.

Сам алгоритм таков:
1) на вход поступает набор двумерных точек
2) создается двумерный массив (аккумулятор) из нулей, размер его соответствует Pi / угловое разрешение и диагональ входного изображения / пространственное разрешение
3) для каждой точки входных данных рассматриваются всевозможные прямые с шагом по углу, равным угловому разрешению
4) каждая такая прямая представляется в виде угла и длины перпендикуляра, опущенного из угла изображения
5) к элементу аккумулятора, соответствующий этой прямой, прибавляется 1
6) результат - это прямая, соответствующая максимуму в аккумуляторе
